# Test Case 5: Temperature Unit Parsing (C/F to Celsius)

## Objective
Standardize the Temperature column by converting all temperature values into Celsius (°C). If Fahrenheit values are present, convert them to Celsius. If values are already in Celsius, verify and retain them.

## Expected Outcome
- All temperatures are stored in Celsius.
- Invalid values are converted to NaN.
- Temperature data is standardized for further analysis.

```python
print("Before Cleaning")
print(production["Temperature"].head(10))

print("\nData Type:")
print(production["Temperature"].dtype)
```

```python
# Function to convert temperature to Celsius
def convert_to_celsius(temp):

    temp = str(temp).strip().upper()

    if temp.endswith("F"):
        value = float(temp.replace("F", "").strip())
        return round((value - 32) * 5 / 9, 2)

    elif temp.endswith("C"):
        value = float(temp.replace("C", "").strip())
        return value

    else:
        return None

# Apply conversion
production["Temperature"] = production["Temperature"].apply(convert_to_celsius)
```

```python
print("After Cleaning")
print(production["Temperature"].head(10))

print("\nData Type:")
print(production["Temperature"].dtype)
```

```python
print("Missing Temperature Values:")
print(production["Temperature"].isna().sum())

print("\nTemperature Statistics:")
print(production["Temperature"].describe())
```

```python
production.to_csv("production_after_testcase5.csv", index=False)

print("✅ Test Case 5 completed successfully!")
```

# Test Case 7: VIN Format Validation & Masking

## Objective
Validate that each Vehicle Identification Number (VIN) contains exactly 17 characters. Create a masked version of the VIN to protect sensitive information during data export.

## Expected Outcome
- VINs are validated.
- Invalid VINs are identified.
- A masked VIN column is created.

```python
print("Before Validation")
print(production["VIN"].head(10))

print("\nVIN Lengths:")
print(production["VIN"].str.len().head(10))
```

```python
# VIN Validation (Must be 17 characters)

production["VIN_Status"] = production["VIN"].apply(
    lambda x: "Valid" if len(str(x)) == 17 else "Invalid"
)
```

```python
# Mask VIN for secure export
# Example:
# MA1AB100000000001
# becomes
# MA1***********001

production["Masked_VIN"] = production["VIN"].apply(
    lambda x: x[:5] + "*" * (len(x) - 8) + x[-3:]
    if len(str(x)) == 17
    else "INVALID VIN"
)
```

```python
print("After Validation")

print(
    production[
        ["VIN", "VIN_Status", "Masked_VIN"]
    ].head(10)
)
```

```python
print("VIN Status Count:")
print(production["VIN_Status"].value_counts())

print("\nInvalid VIN Records:")
print(
    production[
        production["VIN_Status"] == "Invalid"
    ][["VIN"]]
)
```

```python
production.to_csv("production_after_testcase7.csv", index=False)

print("✅ Test Case 7 completed successfully!")
```

# Test Case 10: Sensor Calibration & Drift Check

## Objective
Validate sensor readings by comparing production sensor data with the calibration master dataset.

The objective is to identify possible sensor drift or calibration issues by checking whether sensor measurements are within the expected calibration limits.

## Validation Parameters
- Tool ID matching with Calibration Master
- Torque value validation
- Temperature value validation
- Pressure value validation

## Expected Outcome
- Production sensor readings are compared against calibration limits.
- Sensors outside the acceptable range are flagged.
- Calibration status is assigned to each production record.
- Cleaned production data is generated for further analysis.

```python
print("Calibration Master Columns:")
print(calibration.columns)

print("\nCalibration Master Sample:")
print(calibration.head())
```

```python
print("Production Sensor Columns:")

print(
    production.columns
)
```

```python
print("Production Tool IDs:")
print(production["Tool_ID"].head(10))

print("\nCalibration Sensor IDs:")
print(calibration["Sensor_ID"].head(10))
```

```python
# Extract numeric ID from Tool_ID and Sensor_ID

production["Sensor_Number"] = (
    production["Tool_ID"]
    .str.extract(r'(\d+)')
)

calibration["Sensor_Number"] = (
    calibration["Sensor_ID"]
    .str.extract(r'(\d+)')
)

print(production[["Tool_ID", "Sensor_Number"]].head())

print(calibration[["Sensor_ID", "Sensor_Number"]].head())
```

```python
production = production.merge(
    calibration[
        ["Sensor_Number", "Calibration_Factor"]
    ],
    on="Sensor_Number",
    how="left"
)

print("Calibration data joined successfully!")

print(
    production[
        ["Tool_ID", "Sensor_Number", "Calibration_Factor"]
    ].head(15)
)
```

```python
production["Calibration_Status"] = production["Calibration_Factor"].apply(
    lambda x: "Missing Calibration" 
    if pd.isna(x)
    else (
        "Normal"
        if 0.95 <= x <= 1.05
        else "Drift Detected"
    )
)

print(
    production[
        [
            "Tool_ID",
            "Calibration_Factor",
            "Calibration_Status"
        ]
    ].head(15)
)
```

```python
print("Calibration Status Summary:")
print(
    production["Calibration_Status"].value_counts()
)
```

```python
print("Final Calibration Check:")

print(
    production[
        [
            "Tool_ID",
            "Calibration_Factor",
            "Calibration_Status"
        ]
    ].head(20)
)

print("\nCalibration Status Summary:")
print(
    production["Calibration_Status"].value_counts()
)
```
