# TEST CASE 2

```python
production = pd.read_csv("updated_production_events_cleaned_T1.csv")
print("Before Cleaning:")
print(production["Timestamp"].head(5))
```

```python
import re

def fix_timestamp(ts):
    ts = str(ts).strip().replace("/", "-")

    match = re.search(r'(\d{2}):(\d{2})', ts)

    if match:
        hour = int(match.group(1))
        minute = int(match.group(2))

        hour += minute // 60
        minute = minute % 60

        hour = hour % 24

        ts = re.sub(
            r'\d{2}:\d{2}',
            f"{hour:02d}:{minute:02d}",
            ts
        )

    return ts
production["Timestamp"] = production["Timestamp"].apply(fix_timestamp)
```

```python
print("\nAfter Cleaning:")
print(production["Timestamp"].head(10))
```

```python
production.to_csv("production_events_cleaned_T2.csv", index=False)

print("\nTimestamp normalization completed successfully!")
```

# Test Case 4

```python
import re
```

```python
sensor = pd.read_csv("sensor_readings.csv")
```

```python
print("Before Cleaning:")
print(sensor["Torque"].head(10))
```

```python
def convert_to_nm(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()
    
    match = re.search(r'([\d.]+)\s*(nm|nmm|lb-ft)', value)

    if not match:
        return np.nan

    num = float(match.group(1))
    unit = match.group(2)

    if unit == "nm":
        return num
    elif unit == "nmm":
        return num / 1000
    elif unit == "lb-ft":
        return num * 1.35582
```

```python
sensor["Torque_Nm"] = sensor["Torque"].apply(convert_to_nm)

sensor["Torque_Status"] = sensor["Torque_Nm"].apply(
    lambda x: "Valid" if pd.notna(x) else "Invalid"
)
```

```python
print("\nAfter Cleaning:")
print(sensor[["Torque", "Torque_Nm", "Torque_Status"]].head(10))

sensor.to_csv("sensor_readings_cleaned_T4.csv", index=False)

print("\nTorque cleaning completed successfully!")
```

# Test case 5

```python
sensor = pd.read_csv("sensor_readings_cleaned_T4.csv")
```

```python
print("Before Cleaning:")
print(sensor["Temperature"].head(10))
```

```python
def convert_to_celsius(temp):
    if pd.isna(temp):
        return np.nan

    temp = str(temp).strip().upper()

    num = re.findall(r"\d+\.?\d*", temp)

    if not num:
        return np.nan

    num = float(num[0])

    if "F" in temp:
        return round((num - 32) * 5 / 9, 2)

    elif "C" in temp:
        return num

    else:
        return num
```

```python
sensor["Temperature_C"] = sensor["Temperature"].apply(convert_to_celsius)

sensor["Temperature_Status"] = sensor["Temperature_C"].apply(
    lambda x: "Valid" if pd.notna(x) else "Invalid"
)
```

```python
print("\nAfter Cleaning:")
print(sensor[["Temperature", "Temperature_C", "Temperature_Status"]].head(10))

sensor.to_csv("sensor_readings_cleaned_T5.csv", index=False)

print("\nTemperature conversion completed successfully!")
```

# Test Case 10

```python

sensor = pd.read_csv("sensor_readings_cleaned_T5.csv")
calibration = pd.read_csv("calibration_master.csv")

print("Sensor Readings:")
print(sensor.head())

print("\nCalibration Data:")
print(calibration.head())

```

```python
sensor = sensor.merge(
    calibration[["Sensor_ID", "Calibration_Factor"]],
    on="Sensor_ID",
    how="left"
)
```

```python
print(sensor.columns)
print(calibration.columns)
```

```python
sensor["Torque"] = (
    sensor["Torque"]
    .astype(str)
    .str.replace("Nm", "", case=False, regex=False)
    .str.replace("NM", "", regex=False)
    .str.strip()
)

sensor["Torque"] = pd.to_numeric(sensor["Torque"], errors="coerce")

sensor["Calibration_Factor_y"] = pd.to_numeric(
    sensor["Calibration_Factor_y"],
    errors="coerce"
)
```

```python
sensor["Calibration_Factor_y"] = sensor["Calibration_Factor_y"].fillna(1.0)

sensor["Calibrated_Torque"] = (
    sensor["Torque"] * sensor["Calibration_Factor_y"]
).round(2)

sensor["Drift"] = (
    sensor["Calibrated_Torque"] - sensor["Torque"]
).round(2)

sensor["Calibration_Status"] = sensor["Drift"].apply(
    lambda x: "Within Limit" if abs(x) <= 2 else "Out of Calibration"
)
```

```python
sensor.to_csv("sensor_readings_cleaned_T10.csv", index=False)

print("\nSensor calibration and drift check completed successfully!")
```
